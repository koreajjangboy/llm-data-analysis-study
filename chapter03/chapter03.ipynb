```python
from pathlib import Path


def find_project_root(start: Path) -> Path:
    start = start.resolve()

    for candidate in [start, *start.parents]:
        if (
            (candidate / "data" / "raw").exists()
            and (candidate / "book").exists()
        ):
            return candidate

    raise FileNotFoundError(
        "프로젝트 루트를 찾지 못했습니다. "
        "Notebook을 저장소 안에서 실행해 주세요."
    )


PROJECT_ROOT = find_project_root(Path.cwd())
DATA_DIR = PROJECT_ROOT / "data" / "raw"

print(PROJECT_ROOT)
print(DATA_DIR)
```

📊 customers.csv 구조 요약
행/열: 150행, 6열 (150, 6)

컬럼: customer_id, name, gender, age, city, signup_date

데이터 타입:

name: object (문자열)

gender: object (문자열)

age: int64 (정수형)

city: object (문자열)

결측치: 없음 (모든 컬럼 0건)

전체 중복: 없음 (완전 일치하는 행 0건)

customer_id 중복: 없음 (고유 식별자 중복 0건)

🔍 분석 전 추가 확인 사항 (구분)
데이터 분석을 본격적으로 시작하기 앞서, 단순한 추정(가정)과 데이터에서 실제 확인이 필요한 항목을 구분하여 점검해야 합니다.

1. 실제 확인이 필요한 항목 (데이터 검증)
날짜 타입 변환 여부 (signup_date): 현재 object 타입으로 되어 있으므로, 시계열 분석이나 기간별 필터링을 위해 datetime 타입으로 변환이 필요한지 확인해야 합니다.

값의 유효성 범위 점검:

age(나이): 최소 19세에서 최대 69세로 확인되나, 비정상적인 값(음수나 터무니없는 고령 등)이 숨어있지 않은지 분포를 상세히 확인해야 합니다.

gender(성별): 데이터상 ['F', 'M'] 값만 존재하지만, 공백이나 대소문자 혼용 여부를 명확히 체크해야 합니다.

city(도시): 총 10개의 도시가 포함되어 있으며, 표기 방식(띄어쓰기, 오탈자 등)에 일관성이 있는지 확인해야 합니다.

2. 추정 및 비즈니스 맥락 확인 항목 (추가 정보 필요)
데이터의 수집 기준 및 범위: 이 150명의 고객 데이터가 전체 고객을 대표하는 표본인지, 특정 기간이나 특정 이벤트로 유입된 코호트인지 추정하지 말고 정의를 명확히 해야 합니다.

외래키(FK) 연관성: 앞서 다룬 orders나 order_items 테이블과 조인(Join)할 때, customer_id가 누락 없이 완벽하게 매핑되는지(orders의 customer_id가 본 파일의 ID 범주 내에 완전히 속하는지) 실제 교차 검증이 필요합니다.

```python
def check_data(name, df):
    print(f"{name} 구조 요약")
    print(f"행/열: {df.shape[0]} / {df.shape[1]}")
    print("컬럼:", list(df.columns))
    print("데이터 타입:")
    print(df.dtypes)
    print("결측치:")
    print(df.isna().sum())
    print("customer_id 중복:", df.duplicated(subset=["customer_id"]).sum())

check_data("customers.csv", customers)
```
